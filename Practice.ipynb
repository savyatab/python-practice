```python
# 2.Verify if a given four-digit number is a palindrome Hint: A number is a palindrome if it reads the same forward and backward.
num =input("Enter a 4-digit number: ")
if len(num) == 4:
  if num == num[::-1]:
    print("Palindrome")
  else:
    print("Not Palindrome")
else:
    print("Invalid input")
```

```python
# 4.Classify a triangle based on angles (Acute, Right, Obtuse) Hint: Use the properties of angles in a triangle.
angle1 = int(input("Enter first angle: "))
angle2 = int(input("Enter second angle: "))
angle3 = int(input("Enter third angle: "))
total_angle = angle1 + angle2 + angle3
if total_angle == 180 and angle1 > 0 and angle2 > 0 and angle3 > 0:
  if angle1 == 90 or angle2 == 90 or angle3 == 90:
    print("Right triangle")
  elif angle1 > 90 or angle2 > 90 or angle3 > 90:
   print("Obtuse triangle")
  else:
   print("Acute triangle")
else:
    print("Invalid triangle")
```

```python
# 6. Determine whether a given time represents morning, afternoon, evening, or night Hint: Check different time ranges
hour = int(input('Enter the hour[0-23] :'))
if 5 <= hour < 12:
  print('Morning')
elif 12 <= hour < 17:
  print('Afternoon')
elif 17 <= hour < 21:
  print('Evening')
else:
  print('Night')
```
