```python
#wap of above code using parameters
# Dictionary for mapping words to digits
def num_converter(number_words):
  digits_mapping = {
    'zero': '0',
    'one': '1',
    'two': '2',
    'three': '3',
    'four': '4',
    'five': '5',
    'six': '6',
    'seven': '7',
    'eight': '8',
    'nine': '9'
  }
  words = number_words.lower().split()
  output = ''
  for word in words:
    output += digits_mapping.get(word, '_')
  print(output)
number=input('Please enter your phone number :')
num_converter(number)
```

```python
#why do we need decortors?
from time import time
def time_taken(fn):
  def wrapper(*args,**kwargs):
    t1=time()
    result= fn(*args, **kwargs)
    t2=time()
    print(f'took {t2-t1}seconds')
    return result
  return wrapper

@time_taken
def check_time():
  for i in range(100000):
   i*5
check_time()
```
